```python
vendor_df = pd.read_sql_query("SELECT * FROM vendor_invoice",conn)
vendor_df.head()
'''we are trying to fill the approval section through classification'''
```

```python
'''for estimating freight cost, we first need to find out if there is any corelation between quantity, dollars and freight.'''
vendor_df[['Quantity', 'Dollars', 'Freight']].corr()
```

```python
plt.figure(figsize=(4,2))
sns.heatmap(vendor_df[['Quantity', 'Dollars', 'Freight']].corr(),annot=True)
plt.show()

plt.scatter(vendor_df['Quantity'], vendor_df['Freight'])
plt.scatter(vendor_df['Dollars'], vendor_df['Freight'])
plt.legend(['Quantity', 'Dollars'])
plt.ylabel('Freight Cost')
plt.show()
```

```python
vendor_df['Freight_per_unit'] = vendor_df['Freight'] / vendor_df['Quantity']
vendor_df
```

```python
'''lets see the spread of the dataset,  checking the boundaries for "low" and "high" values'''
low_quantity = vendor_df['Quantity'].quantile(0.25)
high_quantity = vendor_df['Quantity'].quantile(0.75)
```

```python
'''It means that exactly 25% of all the rows in our Quantity column have a quantity of 83.0 or less'''
low_quantity
```

```python
'''It means that 75% of the rows have a quantity of 5100.5 or less. Only 25% of our data has a quantity higher than this.'''
high_quantity
```

```python
'''when quantity is below 83, the mean of Freight/unit is $0.09 '''

round(vendor_df.loc[vendor_df['Quantity'] < low_quantity, 'Freight_per_unit'].mean(), 2)
```

```python
'''when quantity is below 5100.5, the mean of Freight/unit is $0.09'''
'''we can colclude, when we buy in bulk, we pay less Freight/unit ($0.05) than when buying less quantity'''

round(vendor_df.loc[vendor_df['Quantity'] > high_quantity, 'Freight_per_unit'].mean(), 2) 
```

```python
'''we can see from 25% and max(and also from our scatter plot earlier) that there are outliers in our data'''
'''but we can't just remove them as we want to predict freight cost for bulk orderers also. Lets see what happens if we dont remove anything first'''

vendor_df.describe().round()
```

```python
X = vendor_df[['Dollars']]
Y = vendor_df['Freight']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
```

```python
model1 = LinearRegression()
model1.fit(X_train,Y_train)

model2 = DecisionTreeRegressor(max_depth=4, random_state = 42)
model2.fit(X_train,Y_train)

model3 = RandomForestRegressor(max_depth=4, random_state = 42)
model3.fit(X_train,Y_train)
```

```python
def evaluate_model(model, X_test, Y_test, model_name):
    preds = model.predict(X_test)

    mae = mean_absolute_error(Y_test, preds)
    rmse =root_mean_squared_error(Y_test,preds)
    r2 = r2_score(Y_test, preds)*100

    print(f"\n{model_name} Performance: ")
    print(f"MAE  : {mae:.2f}")
    print(f"RMSE : {rmse:.2f}")
    print(f"R²   : {r2:.2f}%")
```

```python
evaluate_model(model1, X_test, Y_test, 'Linear Regression')
evaluate_model(model2, X_test, Y_test, 'Decision Tree Regression')
evaluate_model(model3, X_test, Y_test, 'Random Forest Regression')
```
